```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch

# 저장된 모델 경로
output_dir = "./koalpaca_save"  # 폴더 내 모델 배치

# 4bit 양자화 설정
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

# 모델과 토크나이저 불러오기 (4bit 양자화 설정 포함)
model = AutoModelForCausalLM.from_pretrained(output_dir, quantization_config=bnb_config, device_map="auto")
tokenizer = AutoTokenizer.from_pretrained(output_dir)

# 모델을 GPU로 이동 (필요 시)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print("4bit 양자화된 모델과 토크나이저가 성공적으로 불러와졌습니다.")
```

```python
model.eval()
model.config.use_cache = False  # silence the warnings. Please re-enable for inference!
torch.cuda.empty_cache()
```

```python
def extract_content(text):
    # "답변: " 이후부터 후처리
    if "답변: " in text:
        text = text.split("답변: ", 1)[1]
    else:
        return None  # "### 답변: "이 없으면 None 반환

    # "<|sep|>"를 기준으로 중간 부분 추출
    if "<|sep|>" in text:
        parts = text.split("<|sep|>")
        if len(parts) > 1:
            text = parts[0].strip()

    if "<|endoftext|>" in text:
        parts = text.split("<|endoftext|>")
        if len(parts) > 1:
            text = parts[0].strip()

    if "꿋" in text:
        parts = text.split("꿋")
        if len(parts) > 1:
            text = parts[0].strip()

    keyphrases = ["개입요인", "개입 요인"]
    for keyphrase in keyphrases:
        if keyphrase in text:

            before_keyphrase, after_keyphrase = text.split(keyphrase, 1)
            intervention_description = after_keyphrase.split("\n", 1)[0]

            return before_keyphrase.strip() + "\n" + keyphrase + intervention_description.strip()


    return text
```

```python
def gen(instruction, input_text):
    # instruction과 input을 이용해 프롬프트 생성
    prompt = f"### 명령어: {instruction}\n\n### 맥락: {input_text}\n\n### 답변:"

    # 프롬프트 길이 계산
    prompt_length = len(tokenizer(prompt)['input_ids'])

    # 생성된 프롬프트를 모델에 전달해 답변 생성
    gened = model.generate(
        **tokenizer(
            prompt,
            return_tensors='pt',
            return_token_type_ids=False
        ).to(device),
        max_new_tokens=1024,
        temperature=0.7,
        num_beams=1,  # Beam Search 사용
        no_repeat_ngram_size=3,
        repetition_penalty=1.2,
        early_stopping=True,
        do_sample=False,  # 샘플링 사용
        eos_token_id=None # tokenizer.eos_token_id
    )

    # 생성된 텍스트 디코딩
    generated_text = tokenizer.decode(gened[0])  # skip_special_tokens=True
    post_text = extract_content(generated_text)

    print(post_text)
    return post_text
```

```python
import torch
import gc
import time
import os
```

```python
instruction = "다음과 같은 상담기록을 보고 요약서를 작성해주세요."

output_dir = "./generated"

for files in df['filename']:
  input_text = df.loc[df['filename'] == files, 'original'].values[0]

  print(files)
  gened = gen(instruction, input_text)

  file_path = os.path.join(output_dir, f"{files}.txt")  # 파일 경로
  with open(file_path, "w", encoding="utf-8") as file:
        file.write(gened)  # 파일 내용 작성
        print(f"File created: {file_path}")

  # 가비지 컬렉션 실행
  gc.collect()

  # GPU 메모리 비우기
  torch.cuda.empty_cache()

  # 20초 동안 멈춤
  time.sleep(25)
```
